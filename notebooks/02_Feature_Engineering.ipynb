```python
df['ticket_text'] = (
    df['subject'].fillna('') + ' ' +
    df['body'].fillna('')
)

df[['subject', 'body', 'ticket_text']].head()
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['ticket_text_clean'] = df['ticket_text'].apply(clean_text)

df[['ticket_text', 'ticket_text_clean']].head()
```

```python
print("Empty tickets:", (df['ticket_text_clean'] == '').sum())
```

```python
print("Category (type):")
print(df['type'].value_counts())

print("\nPriority:")
print(df['priority'].value_counts())

print("\nQueue:")
print(df['queue'].value_counts())
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate ticket texts:", df['ticket_text_clean'].duplicated().sum())
```

```python
conflicts = (
    df.groupby('ticket_text_clean')['type']
      .nunique()
)

print("Texts with multiple categories:", (conflicts > 1).sum())
```

```python
feature_df = df[
    [
        'ticket_text_clean',
        'type',
        'priority',
        'queue',
        'language',
        'subject_length',
        'body_length'
    ]
].copy()

feature_df.head()
```
